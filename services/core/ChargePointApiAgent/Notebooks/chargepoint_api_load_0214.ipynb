```python
# similarly, for getLoad
def getLoad(query_para_dict={}):
    data_section_name = "stationData"
    searchQuery= client.get_type('ns0:sessionSearchdata')() 
    for k, v in query_para_dict.items():
        setattr(searchQuery, k, v)
    response_station = client.service.getLoad(searchQuery)
    # return response_station[data_section_name] #construct_data_session(responses, data_section_name)
    return response_station


# load_data = getLoad({"sessionID": 97885729})
# load_data = getLoad({"stationID": "1:13059121"})
load_data = getLoad({"stationID": "5:15504761"})
# load_data = getLoad({})
load_data
```

```python
# for better usability

def getLoadAPI(sgID=None, stationID=None, sessionID=None):
    # Note: 'Search by either sgID, sessionID or stationID', one and only one of them should be provided.
    searchQuery = {} 
    if sgID is not None:
        searchQuery.update({'sgID': sgID}) 
    if stationID is not None:
        searchQuery.update({'stationID': stationID}) 
    if sessionID is not None:
        searchQuery.update({'sessionID': sessionID}) 

    response = client.service.getLoad(searchQuery)
    
    if response["responseCode"] != "100":
                    # raise Exception(f"API error with response: {response}")
                    print(f"API error with response: {response}.")
                    return None
                
    return response

# Usage examples:
load_data_by_station = getLoadAPI(stationID="5:15504761")
load_data_by_session = getLoadAPI(sessionID=97885729)  # 101268659
load_data_by_sgID = getLoadAPI(sgID=12345)
load_data_by_non = getLoadAPI()

print(load_data_by_station)
print(load_data_by_session)
print(load_data_by_sgID)
print(load_data_by_non)
```

Note: If you use the shedLoad call with the percentShedPerStation or percentShedPerPort parameter on a station that is not delivering any power, the maximum output is set to 0 kW, and charging will not be allowed on that station until the shed period ends or a call to clearShedState is made.

```python
# similarly, for shedLoad
def shedLoad(query_para_dict={}):
    data_section_name = "stationData"
    searchQuery= client.get_type('ns0:sessionSearchdata')() 
    for k, v in query_para_dict.items():
        setattr(searchQuery, k, v)
    response_station = client.service.shedLoad(searchQuery)
    # return response_station[data_section_name] #construct_data_session(responses, data_section_name)
    return response_station


# load_data = getLoad({"sessionID": 97885729})
# load_data = getLoad({"stationID": "1:13059121"})
shed_load_response = shedLoad({"stationID": "5:15504761", "timeInterval": 1, "allowedLoadPerPort": 0, "portNumber": 1})
# load_data = getLoad({})
print(f"{shed_load_response = }")

print(f'{getLoad({"stationID": "5:15504761"}) = }')
```

```python
# similarly, for shedLoad
def clearShedState(query_para_dict={}):
    data_section_name = "stationData"
    searchQuery= client.get_type('ns0:sessionSearchdata')() 
    for k, v in query_para_dict.items():
        setattr(searchQuery, k, v)
    print(f"{searchQuery = }")
    response_station = client.service.clearShedState(query_para_dict)
    # return response_station[data_section_name] #construct_data_session(responses, data_section_name)
    return response_station


# load_data = getLoad({"sessionID": 97885729})
# load_data = getLoad({"stationID": "1:13059121"})
clear_shed_response = clearShedState({"stationID": "5:15504761", })  
# load_data = getLoad({})
print(f"{clear_shed_response = }")

print(f'{getLoad({"stationID": "5:15504761"}) = }')
```
