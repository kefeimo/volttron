```python
import psycopg2
from psycopg2 import sql
```

```python
# Connect to an existing database
conn = psycopg2.connect(
    dbname="testdb",
    user="postgres",
    password="password",  # Replace with your postgres user's password
    host="localhost"
)

# Open a cursor to perform database operations
cur = conn.cursor()

# Create table
cur.execute("""
CREATE TABLE IF NOT EXISTS contacts (
    id SERIAL PRIMARY KEY,
    name VARCHAR(100),
    phone VARCHAR(15),
    email VARCHAR(100)
)
""")

# Insert data
cur.execute("""
INSERT INTO contacts (name, phone, email) VALUES (%s, %s, %s),
(%s, %s, %s)
""", ('Alice', '123-456-7890', 'alice@example.com', 'Bob', '234-567-8901', 'bob@example.com'))

# Query data
cur.execute("SELECT * FROM contacts")
contacts = cur.fetchall()
print("Initial data:")
for contact in contacts:
    print(contact)

# Update data
cur.execute("UPDATE contacts SET phone = %s WHERE name = %s", ('345-678-9012', 'Bob'))

# Query data again to see the update
cur.execute("SELECT * FROM contacts")
updated_contacts = cur.fetchall()
print("Updated data:")
for contact in updated_contacts:
    print(contact)

# Delete data
cur.execute("DELETE FROM contacts WHERE name = %s", ('Alice',))

# Commit changes
conn.commit()

# Close communication with the database
cur.close()
conn.close()
```
